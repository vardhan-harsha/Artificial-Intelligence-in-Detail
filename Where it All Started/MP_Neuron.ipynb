## McCulloch-Pitts Neuron

- The McCulloch-Pitts (MP) neuron, introduced in 1943 by Warren McCulloch and Walter Pitts, is one of the earliest artificial neuron models. It's a simplified mathematical model of how biological neurons work. The MP neuron takes binary inputs (0 or 1), applies weights to them, and produces a binary output based on whether the weighted sum exceeds a threshold. This fundamental model laid the groundwork for modern artificial neural networks and demonstrated that simple neural networks could perform complex logical operations.
